# PathSnap: Step 1 - Exploratory Data Analysis (EDA)
This notebook explores the PathSnap dataset across the  and  partitions, validating dataset health, distributions, and spatial network structures.

```
import sys
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sys.path.insert(0, '../src')
from data_io import PathSnapDataLoader

loader = PathSnapDataLoader('../../Student folder')
print('Partitions available:', loader.list_partitions())
```

```
# Inventory & partition sizes
df_inv = pd.read_csv('../outputs/eda/inventory_summary.csv')
print(df_inv.groupby('partition')[['n_points', 'n_nodes', 'n_arcs', 'n_route_arcs']].describe())
```

```
# Timestamp gap analysis
df_ts = pd.read_csv('../outputs/eda/timestamp_summary.csv')
print('Total points with 1s gap: {:.2f}%'.format(df_ts['count_1s'].sum() / (df_ts['count_1s'].sum() + df_ts['count_other'].sum()) * 100))
print('Any duplicate timestamps across trips?:', (df_ts['n_duplicates'] > 0).any())
```

```
# Route integrity checks
df_route = pd.read_csv('../outputs/eda/route_summary.csv')
print('All route arcs valid:', df_route['all_arcs_valid'].all())
print('Trips with 100% strictly consecutive route arcs:', (df_route['pct_strictly_connected'] == 100.0).sum(), '/', len(df_route))
```

```
# Geographic extents and CRS implications
df_ext = pd.read_csv('../outputs/eda/extent_summary.csv')
print('Global lon range:', df_ext['track_lon_min'].min(), 'to', df_ext['track_lon_max'].max())
print('Global lat range:', df_ext['track_lat_min'].min(), 'to', df_ext['track_lat_max'].max())
```
