# PathSnap: Notebook 03 — Candidate Generation & Radius Justification

Evaluates point-to-reference-arc distances on training data, sweeps candidate radii, and justifies selecting search radius R = 50.0m.

```
import os, sys
from pathlib import Path
sys.path.insert(0, str(Path("../src").resolve()))
os.environ["MPLCONFIGDIR"] = "/tmp/matplotlib"

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## 1. Candidate Search Radius Sweep Results

```
sweep_df = pd.read_csv('../outputs/csv/candidate_radius_sweep.csv')
sweep_df
```

## 2. Candidate Recall vs Search Radius Curve

```
plt.figure(figsize=(8, 4.5))
plt.plot(sweep_df['radius_meters'], sweep_df['recall'] * 100, marker='o', color='#1f77b4', linewidth=2)
plt.axvline(50.0, color='green', linestyle='--', label='Selected R = 50.0m')
plt.title('Candidate Recall (%) vs Search Radius (m)')
plt.xlabel('Search Radius R (m)')
plt.ylabel('Recall (%)')
plt.ylim(75, 101)
plt.legend()
plt.tight_layout()
plt.show()
```
